### **Estimation of number of runs**

```python
# Common values
# z for 95% confidence 
z=1.96

# Initial withdrawal rate to 4%
est_withdrawal_rate = 0.04

# 500 number of runs
est_runs = 500
```

### Full Home

```python
# Run Initial MCS
czech, european, american = 1, 0, 0
est_fh_portfolio, est_fg_withdrawal = mcs(runs=est_runs, withdrawal_rate=est_withdrawal_rate, cz=czech, eu=european, us=american)
```

```python
# Obtain final values of simulated portfolios
fh_final=[]
for i in range(len(est_fh_portfolio)):
    if len(est_fh_portfolio[i])==31:
        fh_final.append(est_fh_portfolio[i][-1])
    else:
        fh_final.append(0)

# Calculate statistics
est_fh_mean = np.mean(fh_final)
est_fh_std = np.std(fh_final)

# Calculate n
#n_full_home = ((z * est_fh_std) / (est_fh_mean * 0.01))**2
n_full_home = ((z * est_fh_std) / (300_000))**2
print(f"The error rate in a Full Home scenario is set to {np.round(est_fh_mean * 0.01,2)} and the number of runs should be at least {np.ceil(n_full_home)}.")
```
