## **Inflation pre-processing**

This jupyter notebook is used to pre-process data on inflation.



```python
# Import of necessary packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as st
```

```python
# Plot between -10 and 10 with .001 steps.
x_axis = np.arange(-10, 10, 0.0001)
#x_axis = np.arange(-0.08, 0.05, 0.00001)
# Mean = 0, SD = 2.
plt.plot(x_axis, st.norm.pdf(x_axis,0.06, 0.052530572))
plt.show()
```
